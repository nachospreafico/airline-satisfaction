```python
import pandas as pd
import numpy as np
```

```python
from pathlib import Path

data_path = Path("../data/raw/train.csv")
df = pd.read_csv(data_path)
```

```python
# Shape of the dataset
rows, cols = df.shape
print(f"Dataset contains {rows} rows and {cols} columns")
```

```python
# Column names, their dtypes and count of non-null values
print("Dataset contains following columns:")
df.info()
```

```python
# First 5 rows of the dataset
df.head()
```

```python
# Class counts and percentages (check imbalance)
class_check = df.groupby("satisfaction", as_index=False).agg(count=("id", "count"))
class_check["pct"] = class_check["count"] / class_check["count"].sum() * 100
class_check
```

```python
# Check for missing values
missing_values_df = df.isna().sum().to_frame()
missing_values_df = missing_values_df.rename(columns={0: "count_of_missing_values"})
missing_values_df["pct"] = missing_values_df["count_of_missing_values"] / rows * 100
missing_values_df
```

```python
# Check for duplicated rows (excluding 'id' column)
df[[col for col in df.columns if col != "id"]].duplicated().sum()
```

```python
# Summary statistics
df.describe()
```

```python
# Categorical features: Unique values and frequency counts
categorical_columns = ["Gender", "Customer Type", "Type of Travel", "Class"]
for col in categorical_columns:
    print(f"Unique values and frequency of column {col}:")
    print(df[col].value_counts())
    print()
```

```python
# Convert target column (satisfaction) into int (True = 1, False = 0)
df["satisfaction"] = df["satisfaction"].astype(int)
```

```python
# Investigate correlation between numerical variables and target variable
numerical_cols = [col for col in df.columns if col not in categorical_columns]
abs(df[numerical_cols].corr()["satisfaction"]).sort_values(ascending=False)
```
